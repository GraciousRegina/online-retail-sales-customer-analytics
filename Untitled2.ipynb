## Phase 2: Data Cleaning, Feature Engineering, and KPI Development

This phase prepares the Online Retail dataset for reliable business analysis.
It examines data quality, removes duplicate and invalid sales records,
preserves missing customer identifiers where appropriate, engineers
time-based and revenue features, and computes key sales performance indicators.

Cancelled transactions and returns are excluded from the completed-sales
dataset and retained separately for future analysis.

```python

# ============================================================
# PHASE 2: DATA CLEANING, FEATURE ENGINEERING & SALES KPIs
# Online Retail Sales Analysis and Dashboard Project
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

print("=" * 65)
print("PHASE 2: DATA CLEANING & BUSINESS KPI DEVELOPMENT")
print("=" * 65)

# ------------------------------------------------------------
# 2.1 PRESERVE ORIGINAL DATA
# ------------------------------------------------------------

raw_df = df.copy()

print("\n[1] ORIGINAL DATASET")
print(f"Rows: {len(raw_df):,}")
print(f"Columns: {raw_df.shape[1]}")
print(f"Duplicate rows: {raw_df.duplicated().sum():,}")

print("\nMissing values:")
display(raw_df.isnull().sum().to_frame("Missing Count"))

# ------------------------------------------------------------
# 2.2 STANDARDIZE DATA TYPES
# ------------------------------------------------------------

retail = raw_df.copy()

retail["InvoiceNo"] = retail["InvoiceNo"].astype(str).str.strip()
retail["StockCode"] = retail["StockCode"].astype(str).str.strip()

retail["Description"] = (
    retail["Description"]
    .astype("string")
    .str.strip()
)

retail["Country"] = retail["Country"].astype(str).str.strip()

retail["InvoiceDate"] = pd.to_datetime(
    retail["InvoiceDate"],
    errors="coerce"
)

retail["Quantity"] = pd.to_numeric(
    retail["Quantity"], errors="coerce"
)

retail["UnitPrice"] = pd.to_numeric(
    retail["UnitPrice"], errors="coerce"
)

retail["CustomerID"] = pd.to_numeric(
    retail["CustomerID"], errors="coerce"
).astype("Int64")

print("\n[2] DATA TYPES STANDARDIZED")
print(retail.dtypes)

# ------------------------------------------------------------
# 2.3 IDENTIFY TRANSACTION QUALITY ISSUES
# ------------------------------------------------------------

retail["IsCancelled"] = (
    retail["InvoiceNo"].str.upper().str.startswith("C")
)

quality_report = {
    "Original Records": len(retail),
    "Duplicate Records": int(retail.duplicated(
        subset=raw_df.columns.tolist()
    ).sum()),
    "Missing Descriptions": int(retail["Description"].isna().sum()),
    "Missing Customer IDs": int(retail["CustomerID"].isna().sum()),
    "Invalid Dates": int(retail["InvoiceDate"].isna().sum()),
    "Nonpositive Quantities": int((retail["Quantity"] <= 0).sum()),
    "Nonpositive Prices": int((retail["UnitPrice"] <= 0).sum()),
    "Cancelled Invoice Lines": int(retail["IsCancelled"].sum())
}

print("\n[3] DATA QUALITY REPORT")
display(pd.DataFrame(
    quality_report.items(),
    columns=["Quality Metric", "Count"]
))

# ------------------------------------------------------------
# 2.4 PRESERVE CANCELLATIONS AND RETURNS
# ------------------------------------------------------------

returns_df = retail[
    retail["IsCancelled"] |
    (retail["Quantity"] < 0)
].copy()

print("\n[4] RETURNS / CANCELLATIONS")
print(f"Flagged transaction lines: {len(returns_df):,}")

# ------------------------------------------------------------
# 2.5 CLEAN COMPLETED SALES
# ------------------------------------------------------------

retail = retail.drop_duplicates(
    subset=raw_df.columns.tolist()
)

retail = retail.dropna(
    subset=["InvoiceDate", "Quantity", "UnitPrice", "Description"]
)

retail = retail[
    retail["Description"].str.len().fillna(0) > 0
].copy()

retail = retail[
    (retail["Quantity"] > 0) &
    (retail["UnitPrice"] > 0) &
    (~retail["IsCancelled"])
].copy()

# CustomerID missing values are preserved because
# these transactions still contribute to revenue.

retail = retail.drop(columns=["IsCancelled"])
retail = retail.reset_index(drop=True)

print("\n[5] CLEANED SALES DATASET")
print(f"Cleaned rows: {len(retail):,}")
print(f"Removed rows: {len(raw_df) - len(retail):,}")
print(f"Missing CustomerIDs retained: {retail['CustomerID'].isna().sum():,}")

# ------------------------------------------------------------
# 2.6 FEATURE ENGINEERING
# ------------------------------------------------------------

retail["TotalRevenue"] = (
    retail["Quantity"] * retail["UnitPrice"]
)

retail["Date"] = retail["InvoiceDate"].dt.date
retail["Year"] = retail["InvoiceDate"].dt.year
retail["Month"] = retail["InvoiceDate"].dt.month
retail["MonthName"] = retail["InvoiceDate"].dt.month_name()

retail["MonthYear"] = (
    retail["InvoiceDate"].dt.to_period("M").astype(str)
)

retail["Quarter"] = retail["InvoiceDate"].dt.quarter
retail["DayOfWeek"] = retail["InvoiceDate"].dt.day_name()
retail["Hour"] = retail["InvoiceDate"].dt.hour

retail["IsWeekend"] = (
    retail["InvoiceDate"].dt.dayofweek >= 5
)

print("\n[6] ENGINEERED FEATURES")
print(retail.columns.tolist())

# ------------------------------------------------------------
# 2.7 CALCULATE BUSINESS KPIs
# ------------------------------------------------------------

total_revenue = retail["TotalRevenue"].sum()
total_orders = retail["InvoiceNo"].nunique()
total_units = retail["Quantity"].sum()
unique_customers = retail["CustomerID"].nunique()
unique_products = retail["StockCode"].nunique()
countries = retail["Country"].nunique()

average_order_value = (
    total_revenue / total_orders if total_orders else np.nan
)

average_units_per_order = (
    total_units / total_orders if total_orders else np.nan
)

kpi_table = pd.DataFrame({
    "KPI": [
        "Total Sales Revenue",
        "Total Orders",
        "Total Units Sold",
        "Identified Customers",
        "Unique Products",
        "Countries Served",
        "Average Order Value",
        "Average Units per Order"
    ],
    "Value": [
        f"£{total_revenue:,.2f}",
        f"{total_orders:,}",
        f"{total_units:,}",
        f"{unique_customers:,}",
        f"{unique_products:,}",
        f"{countries:,}",
        f"£{average_order_value:,.2f}",
        f"{average_units_per_order:,.2f}"
    ]
})

print("\n[7] KEY BUSINESS PERFORMANCE INDICATORS")
display(kpi_table)

# ------------------------------------------------------------
# 2.8 MONTHLY SALES AGGREGATION
# ------------------------------------------------------------

monthly_sales = (
    retail.groupby("MonthYear", as_index=False)
    .agg(
        Revenue=("TotalRevenue", "sum"),
        Orders=("InvoiceNo", "nunique"),
        UnitsSold=("Quantity", "sum")
    )
    .sort_values("MonthYear")
)

print("\n[8] MONTHLY SALES SUMMARY")
display(monthly_sales)

# ------------------------------------------------------------
# 2.9 TOP PRODUCTS BY REVENUE
# ------------------------------------------------------------

top_products = (
    retail.groupby(["StockCode", "Description"], as_index=False)
    .agg(
        Revenue=("TotalRevenue", "sum"),
        UnitsSold=("Quantity", "sum")
    )
    .sort_values("Revenue", ascending=False)
)

print("\n[9] TOP 10 PRODUCTS BY REVENUE")
display(top_products.head(10))

# ------------------------------------------------------------
# 2.10 COUNTRY-LEVEL SALES SUMMARY
# ------------------------------------------------------------

country_sales = (
    retail.groupby("Country", as_index=False)
    .agg(
        Revenue=("TotalRevenue", "sum"),
        Orders=("InvoiceNo", "nunique"),
        UnitsSold=("Quantity", "sum")
    )
    .sort_values("Revenue", ascending=False)
)

print("\n[10] TOP 10 COUNTRIES BY REVENUE")
display(country_sales.head(10))

# ------------------------------------------------------------
# 2.11 FINAL VALIDATION
# ------------------------------------------------------------

assert retail["Quantity"].gt(0).all()
assert retail["UnitPrice"].gt(0).all()
assert retail["TotalRevenue"].gt(0).all()
assert retail["InvoiceDate"].notna().all()
assert not retail["InvoiceNo"].str.upper().str.startswith("C").any()

print("\n[11] FINAL DATA VALIDATION")
print("Date range:", retail["InvoiceDate"].min(),
      "to", retail["InvoiceDate"].max())
print("Final shape:", retail.shape)
print("Total missing values:", retail.isna().sum().sum())

# ------------------------------------------------------------
# 2.12 SAVE CLEANED DATA AND KPI TABLE
# ------------------------------------------------------------

retail.to_csv("OnlineRetail_Cleaned.csv", index=False)
kpi_table.to_csv("Retail_KPI_Summary.csv", index=False)
monthly_sales.to_csv("Retail_Monthly_Sales.csv", index=False)

print("\nFiles saved successfully.")
print("OnlineRetail_Cleaned.csv")
print("Retail_KPI_Summary.csv")
print("Retail_Monthly_Sales.csv")

print("\n" + "=" * 65)
print("PHASE 2 COMPLETED SUCCESSFULLY")
print("=" * 65)
```

## Phase 4: Customer Behavior and RFM Segmentation

This phase investigates customer purchasing behavior using Recency,
Frequency, and Monetary (RFM) analysis. Customers are segmented according
to purchasing activity and transaction value to identify high-value
customers, retention opportunities, and potential customer churn risks.

The analysis also evaluates revenue concentration and customer loyalty.

```python

# ============================================================
# PHASE 4: CUSTOMER BEHAVIOR & RFM SEGMENTATION
# ONLINE RETAIL SALES ANALYTICS PROJECT
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")
os.makedirs("retail_visualizations", exist_ok=True)

print("=" * 65)
print("PHASE 4: CUSTOMER BEHAVIOR & RFM SEGMENTATION")
print("=" * 65)

# ------------------------------------------------------------
# 4.1 PREPARE CUSTOMER DATA
# ------------------------------------------------------------

customer_df = retail.dropna(subset=["CustomerID"]).copy()

print("\nCUSTOMER DATA SUMMARY")
print(f"Customer transaction lines: {len(customer_df):,}")
print(f"Unique customers: {customer_df['CustomerID'].nunique():,}")

# ------------------------------------------------------------
# 4.2 CALCULATE RFM METRICS
# ------------------------------------------------------------

# Use one day after the last transaction as reference date.
snapshot_date = customer_df["InvoiceDate"].max() + pd.Timedelta(days=1)

rfm = (
    customer_df.groupby("CustomerID")
    .agg(
        LastPurchase=("InvoiceDate", "max"),
        Frequency=("InvoiceNo", "nunique"),
        Monetary=("TotalRevenue", "sum")
    )
    .reset_index()
)

rfm["Recency"] = (
    snapshot_date - rfm["LastPurchase"]
).dt.days

rfm = rfm.drop(columns=["LastPurchase"])

print("\nRFM METRICS")
display(rfm.head())

print("\nRFM DESCRIPTIVE STATISTICS")
display(rfm[["Recency", "Frequency", "Monetary"]].describe())

# ------------------------------------------------------------
# 4.3 CREATE RFM SCORES
# ------------------------------------------------------------

# Rank-based quintiles handle repeated frequency values.
# Higher scores represent more favorable customer behavior.

rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first", ascending=False),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["RFM_Score"] = (
    rfm["R_Score"] +
    rfm["F_Score"] +
    rfm["M_Score"]
)

# ------------------------------------------------------------
# 4.4 CUSTOMER SEGMENTATION
# ------------------------------------------------------------

def assign_segment(row):
    r = row["R_Score"]
    f = row["F_Score"]
    m = row["M_Score"]

    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"
    elif r >= 3 and f >= 4:
        return "Loyal Customers"
    elif r >= 4 and f <= 2:
        return "New / Promising"
    elif r <= 2 and f >= 3:
        return "At Risk"
    elif r <= 2 and f <= 2:
        return "Hibernating"
    else:
        return "Potential Loyalists"

rfm["Segment"] = rfm.apply(assign_segment, axis=1)

segment_summary = (
    rfm.groupby("Segment")
    .agg(
        Customers=("CustomerID", "count"),
        TotalRevenue=("Monetary", "sum"),
        AverageSpending=("Monetary", "mean"),
        AverageFrequency=("Frequency", "mean"),
        AverageRecency=("Recency", "mean")
    )
    .reset_index()
)

segment_summary["CustomerShare"] = (
    segment_summary["Customers"] /
    segment_summary["Customers"].sum() * 100
)

segment_summary["RevenueShare"] = (
    segment_summary["TotalRevenue"] /
    segment_summary["TotalRevenue"].sum() * 100
)

segment_summary = segment_summary.sort_values(
    "TotalRevenue", ascending=False
)

print("\nCUSTOMER SEGMENT SUMMARY")
display(segment_summary.round(2))

# ------------------------------------------------------------
# 4.5 VISUALIZE CUSTOMER SEGMENTS
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(
    data=segment_summary,
    x="Customers",
    y="Segment",
    color="#2563EB",
    ax=ax
)

ax.set_title("Customer Distribution by RFM Segment",
             fontweight="bold")
ax.set_xlabel("Number of Customers")
ax.set_ylabel("Customer Segment")

plt.tight_layout()
plt.savefig(
    "retail_visualizations/09_customer_segments.png", dpi=200
)
plt.show()

# ------------------------------------------------------------
# 4.6 REVENUE CONTRIBUTION BY SEGMENT
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(
    data=segment_summary,
    x="RevenueShare",
    y="Segment",
    color="#0D9488",
    ax=ax
)

ax.set_title("Revenue Contribution by Customer Segment",
             fontweight="bold")
ax.set_xlabel("Share of Identified Customer Revenue (%)")
ax.set_ylabel("Customer Segment")

plt.tight_layout()
plt.savefig(
    "retail_visualizations/10_segment_revenue.png", dpi=200
)
plt.show()

# ------------------------------------------------------------
# 4.7 CUSTOMER REVENUE CONCENTRATION
# ------------------------------------------------------------

customer_ranked = rfm.sort_values(
    "Monetary", ascending=False
).reset_index(drop=True)

customer_ranked["CumulativeRevenue"] = (
    customer_ranked["Monetary"].cumsum() /
    customer_ranked["Monetary"].sum() * 100
)

customer_ranked["CumulativeCustomers"] = (
    np.arange(1, len(customer_ranked) + 1) /
    len(customer_ranked) * 100
)

fig, ax = plt.subplots(figsize=(11, 6))

ax.plot(
    customer_ranked["CumulativeCustomers"],
    customer_ranked["CumulativeRevenue"],
    linewidth=2.5,
    color="#7C3AED"
)

ax.plot(
    [0, 100], [0, 100],
    linestyle="--",
    color="gray",
    label="Equal Revenue Distribution"
)

ax.set_title("Customer Revenue Concentration",
             fontweight="bold")
ax.set_xlabel("Cumulative Share of Customers (%)")
ax.set_ylabel("Cumulative Share of Revenue (%)")
ax.legend()

plt.tight_layout()
plt.savefig(
    "retail_visualizations/11_revenue_concentration.png", dpi=200
)
plt.show()

# ------------------------------------------------------------
# 4.8 CUSTOMER PURCHASE FREQUENCY
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(11, 6))

sns.histplot(
    rfm["Frequency"],
    bins=40,
    color="#F59E0B",
    ax=ax
)

ax.set_xlim(0, rfm["Frequency"].quantile(0.99))
ax.set_title("Customer Purchase Frequency Distribution",
             fontweight="bold")
ax.set_xlabel("Number of Unique Orders")
ax.set_ylabel("Customers")

plt.tight_layout()
plt.savefig(
    "retail_visualizations/12_purchase_frequency.png", dpi=200
)
plt.show()

# ------------------------------------------------------------
# 4.9 CUSTOMER RECENCY DISTRIBUTION
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(11, 6))

sns.histplot(
    rfm["Recency"],
    bins=35,
    color="#DC2626",
    ax=ax
)

ax.set_title("Customer Recency Distribution",
             fontweight="bold")
ax.set_xlabel("Days Since Last Purchase")
ax.set_ylabel("Customers")

plt.tight_layout()
plt.savefig(
    "retail_visualizations/13_customer_recency.png", dpi=200
)
plt.show()

# ------------------------------------------------------------
# 4.10 AUTOMATED CUSTOMER INSIGHTS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("CUSTOMER ANALYTICS INSIGHTS")
print("=" * 65)

print(f"Total identified customers: {len(rfm):,}")

print(
    f"Average customer spending: "
    f"£{rfm['Monetary'].mean():,.2f}"
)

print(
    f"Median customer spending: "
    f"£{rfm['Monetary'].median():,.2f}"
)

print(
    f"Average purchase frequency: "
    f"{rfm['Frequency'].mean():.2f}"
)

print(
    f"Average customer recency: "
    f"{rfm['Recency'].mean():.2f} days"
)

top_20_count = max(1, int(np.ceil(len(rfm) * 0.20)))

top_20_revenue = (
    customer_ranked.head(top_20_count)["Monetary"].sum() /
    customer_ranked["Monetary"].sum() * 100
)

print(
    f"Revenue generated by top 20% of identified customers: "
    f"{top_20_revenue:.2f}%"
)

print("\nCUSTOMER SEGMENT COUNTS")
print(rfm["Segment"].value_counts())

# ------------------------------------------------------------
# 4.11 EXPORT CUSTOMER ANALYTICS
# ------------------------------------------------------------

rfm.to_csv("Retail_RFM_Segmentation.csv", index=False)

segment_summary.to_csv(
    "Retail_Customer_Segment_Summary.csv", index=False
)

print("\nCustomer segmentation files saved successfully.")
print("5 additional visualizations generated.")

print("\n" + "=" * 65)
print("PHASE 4 COMPLETED SUCCESSFULLY")
print("=" * 65)
```

## Phase 5: Interactive Retail Sales and Customer Analytics Dashboard

This phase integrates sales performance, geographic trends, product
analytics, and RFM customer segmentation into an interactive business
intelligence dashboard using Plotly and ipywidgets.

The dashboard supports dynamic filtering by country and transaction
date, allowing users to explore key business metrics and patterns.

```python

# ============================================================
# PHASE 5: INTERACTIVE RETAIL SALES ANALYTICS DASHBOARD
# ============================================================

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import ipywidgets as widgets
from IPython.display import display, clear_output
import os

print("=" * 65)
print("PHASE 5: INTERACTIVE BUSINESS INTELLIGENCE DASHBOARD")
print("=" * 65)

# ------------------------------------------------------------
# 5.1 PREPARE DASHBOARD DATA
# ------------------------------------------------------------

dashboard_df = retail.copy()
dashboard_df["InvoiceDate"] = pd.to_datetime(
    dashboard_df["InvoiceDate"]
)
dashboard_df["CustomerID"] = pd.to_numeric(
    dashboard_df["CustomerID"], errors="coerce"
).astype("Int64")

dashboard_df["MonthPeriod"] = (
    dashboard_df["InvoiceDate"].dt.to_period("M")
)

# Add customer segments from Phase 4.
# Missing CustomerIDs remain unclassified.

segment_lookup = rfm[["CustomerID", "Segment"]].copy()

dashboard_df = dashboard_df.merge(
    segment_lookup,
    on="CustomerID",
    how="left",
    validate="many_to_one"
)

dashboard_df["Segment"] = dashboard_df["Segment"].fillna(
    "Unidentified Customer"
)

# Exclude service charges from product-only rankings.
service_codes = {
    "POST", "DOT", "M", "D", "BANK CHARGES",
    "AMAZONFEE", "CRUK", "S", "C2", "PADS"
}

dashboard_df["IsMerchandise"] = (
    ~dashboard_df["StockCode"].astype(str)
    .str.upper().isin(service_codes)
)

# ------------------------------------------------------------
# 5.2 DEFINE INTERACTIVE FILTERS
# ------------------------------------------------------------

countries = ["All Countries"] + sorted(
    dashboard_df["Country"].dropna().unique().tolist()
)

country_filter = widgets.Dropdown(
    options=countries,
    value="All Countries",
    description="Country:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="320px")
)

min_date = dashboard_df["InvoiceDate"].min().date()
max_date = dashboard_df["InvoiceDate"].max().date()

start_date = widgets.DatePicker(
    description="Start Date:",
    value=min_date,
    style={"description_width": "initial"}
)

end_date = widgets.DatePicker(
    description="End Date:",
    value=max_date,
    style={"description_width": "initial"}
)

refresh_button = widgets.Button(
    description="Update Dashboard",
    button_style="primary",
    icon="refresh"
)

output = widgets.Output()

# ------------------------------------------------------------
# 5.3 DASHBOARD GENERATION FUNCTION
# ------------------------------------------------------------

def generate_dashboard(country, start, end):

    if start is None or end is None or start > end:
        print("Please select a valid date range.")
        return

    filtered = dashboard_df.copy()

    if country != "All Countries":
        filtered = filtered[filtered["Country"] == country]

    filtered = filtered[
        (filtered["InvoiceDate"].dt.date >= start) &
        (filtered["InvoiceDate"].dt.date <= end)
    ].copy()

    if filtered.empty:
        print("No transactions found for these filters.")
        return

    # --------------------------------------------------------
    # KPI CALCULATIONS
    # --------------------------------------------------------

    revenue = filtered["TotalRevenue"].sum()
    orders = filtered["InvoiceNo"].nunique()
    units = filtered["Quantity"].sum()
    customers = filtered["CustomerID"].nunique()
    avg_order = revenue / orders if orders else 0

    # KPI indicator cards
    fig_kpi = make_subplots(
        rows=1,
        cols=4,
        specs=[[{"type": "indicator"}] * 4]
    )

    kpis = [
        ("Sales Revenue", revenue, "£"),
        ("Total Orders", orders, ""),
        ("Units Sold", units, ""),
        ("Average Order Value", avg_order, "£")
    ]

    for i, (label, value, prefix) in enumerate(kpis):
        fig_kpi.add_trace(
            go.Indicator(
                mode="number",
                value=float(value),
                number={
                    "prefix": prefix,
                    "valueformat": ",.2f" if prefix else ",.0f"
                },
                title={"text": label}
            ),
            row=1,
            col=i + 1
        )

    fig_kpi.update_layout(
        title="RETAIL SALES PERFORMANCE OVERVIEW",
        height=220,
        template="plotly_white",
        margin=dict(t=70, b=20, l=20, r=20)
    )

    fig_kpi.show()

    print(f"Identified customers: {customers:,}")
    print(f"Selected country: {country}")
    print(f"Reporting period: {start} to {end}")

    # --------------------------------------------------------
    # MONTHLY REVENUE TREND
    # --------------------------------------------------------

    monthly = (
        filtered.groupby("MonthPeriod", as_index=False)
        .agg(
            Revenue=("TotalRevenue", "sum"),
            Orders=("InvoiceNo", "nunique")
        )
    )

    monthly["Month"] = monthly["MonthPeriod"].astype(str)

    fig1 = px.line(
        monthly,
        x="Month",
        y="Revenue",
        markers=True,
        title="Monthly Sales Revenue Trend",
        labels={"Revenue": "Revenue (£)"}
    )

    fig1.update_traces(line_width=3)
    fig1.update_layout(template="plotly_white", height=420)
    fig1.show()

    # --------------------------------------------------------
    # TOP PRODUCTS BY REVENUE
    # --------------------------------------------------------

    merchandise = filtered[filtered["IsMerchandise"]]

    top_products = (
        merchandise.groupby(
            ["StockCode", "Description"], as_index=False
        )["TotalRevenue"]
        .sum()
        .nlargest(10, "TotalRevenue")
        .sort_values("TotalRevenue")
    )

    fig2 = px.bar(
        top_products,
        x="TotalRevenue",
        y="Description",
        orientation="h",
        title="Top 10 Merchandise Products by Revenue",
        labels={
            "TotalRevenue": "Revenue (£)",
            "Description": "Product"
        },
        color="TotalRevenue",
        color_continuous_scale="Blues"
    )

    fig2.update_layout(
        template="plotly_white",
        height=500,
        showlegend=False,
        coloraxis_showscale=False
    )
    fig2.show()

    # --------------------------------------------------------
    # GEOGRAPHIC REVENUE
    # --------------------------------------------------------

    country_sales = (
        filtered.groupby("Country", as_index=False)
        ["TotalRevenue"]
        .sum()
        .nlargest(10, "TotalRevenue")
    )

    fig3 = px.bar(
        country_sales,
        x="Country",
        y="TotalRevenue",
        color="TotalRevenue",
        title="Top Countries by Revenue",
        labels={"TotalRevenue": "Revenue (£)"},
        color_continuous_scale="Teal"
    )

    fig3.update_layout(
        template="plotly_white",
        height=430,
        coloraxis_showscale=False
    )
    fig3.show()

    # --------------------------------------------------------
    # CUSTOMER SEGMENT DISTRIBUTION
    # --------------------------------------------------------

    identified = filtered.dropna(subset=["CustomerID"])

    # Count distinct identified customers in each segment.
    segment_counts = (
        identified.groupby("Segment")["CustomerID"]
        .nunique()
        .reset_index(name="Customers")
    )

    if not segment_counts.empty:
        fig4 = px.pie(
            segment_counts,
            names="Segment",
            values="Customers",
            hole=0.5,
            title="RFM Customer Segment Distribution"
        )

        fig4.update_layout(
            template="plotly_white",
            height=460
        )
        fig4.show()

    # --------------------------------------------------------
    # REVENUE BY DAY OF WEEK
    # --------------------------------------------------------

    day_order = [
        "Monday", "Tuesday", "Wednesday",
        "Thursday", "Friday", "Saturday", "Sunday"
    ]

    weekday = (
        filtered.groupby("DayOfWeek")["TotalRevenue"]
        .sum()
        .reindex(day_order, fill_value=0)
        .reset_index()
    )

    fig5 = px.bar(
        weekday,
        x="DayOfWeek",
        y="TotalRevenue",
        title="Sales Revenue by Day of Week",
        color="TotalRevenue",
        color_continuous_scale="Purples",
        labels={
            "DayOfWeek": "Day",
            "TotalRevenue": "Revenue (£)"
        }
    )

    fig5.update_layout(
        template="plotly_white",
        height=420,
        coloraxis_showscale=False
    )
    fig5.show()

    # --------------------------------------------------------
    # HOURLY ORDER ACTIVITY
    # --------------------------------------------------------

    hourly = (
        filtered.groupby("Hour")["InvoiceNo"]
        .nunique()
        .reindex(range(24), fill_value=0)
        .reset_index(name="Orders")
    )

    fig6 = px.line(
        hourly,
        x="Hour",
        y="Orders",
        markers=True,
        title="Hourly Customer Ordering Activity"
    )

    fig6.update_layout(
        template="plotly_white",
        height=400
    )
    fig6.show()

    # --------------------------------------------------------
    # SUMMARY INSIGHTS
    # --------------------------------------------------------

    print("\n" + "=" * 65)
    print("DASHBOARD BUSINESS INSIGHTS")
    print("=" * 65)

    print(f"Total revenue: £{revenue:,.2f}")
    print(f"Total orders: {orders:,}")
    print(f"Average order value: £{avg_order:,.2f}")

    if not monthly.empty:
        peak = monthly.loc[monthly["Revenue"].idxmax()]
        print(f"Highest revenue month: {peak['Month']}")

    if not country_sales.empty:
        print(
            f"Leading country: "
            f"{country_sales.iloc[0]['Country']}"
        )

    if not top_products.empty:
        print(
            f"Leading merchandise product: "
            f"{top_products.iloc[-1]['Description']}"
        )

    print(
        "\nNote: RFM segments are based on the full historical "
        "dataset and are not recalculated for selected filters."
    )

    print(
        "Revenue represents positive completed-sales lines, "
        "not net revenue after returns."
    )

# ------------------------------------------------------------
# 5.4 CONNECT DASHBOARD CONTROLS
# ------------------------------------------------------------

def update_dashboard(button=None):
    with output:
        clear_output(wait=True)

        generate_dashboard(
            country_filter.value,
            start_date.value,
            end_date.value
        )

refresh_button.on_click(update_dashboard)

# ------------------------------------------------------------
# 5.5 DISPLAY DASHBOARD
# ------------------------------------------------------------

print("\nINTERACTIVE DASHBOARD CONTROLS")

display(
    widgets.VBox([
        widgets.HBox([country_filter, refresh_button]),
        widgets.HBox([start_date, end_date])
    ])
)

display(output)

# Initial dashboard rendering
update_dashboard()

print("\nPHASE 5 DASHBOARD INITIALIZED SUCCESSFULLY")
```
